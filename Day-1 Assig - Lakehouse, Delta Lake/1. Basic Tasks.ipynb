<!-- warehouse, a lake, and a lakehouse -->
1. warehouse -> data warehouse is place where we store only strucutred data like tables , relational data bases which is cleaned , organised and structure.
-> ware house are quite expensive rather than lake and lake house .
-> ware house allow us to fast query the result from data since the data is strucutred.
-> ware house are consistent and more suitable for enterprice real time data .
-> can't store the unstructure data like pdf,docs,img,notes etc...

2. data lake -> data lake is actually like the lake for the data. where we can store anything like structure data , unstrucutre data and semi structure data.
-> basically data lake keep all raw data , nulls , pdf , docs , csv , json , parquet etc .
-> data lake is not suggested for fast query result since data stored in messay way.
-> data lake are quite cheaper rather than warehouse and lake house 
-> mostly save the data required for data engineer and analytics not for real time use.

3. lake house -> in one word can say  lakehouse = data lake + data warehouse
-> lake house is the combination of lake and warehouse.
-> it uses the parquet file formate with meta data layer on top of it to store the big data.
-> high query result for structure data.
-> provide the space for unstructure data like lake house.
-> it is cheap , managed , flexible with the high ACID properties
->example - delta lake


```sql
-- 2. Create a Delta table from ~10 rows of sample product data (product_id, name, category, price). 
CREATE TABLE dev.bronze.products(
    product_id INT,
    name STRING,
    category STRING,
    price DECIMAL(10,2)
    )
USING DELTA;
```

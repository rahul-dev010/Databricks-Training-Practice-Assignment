```sql
%python
# Method 1: Using mergeSchema option in DataFrame API
from pyspark.sql import Row

# Create a DataFrame with a new column (supplier_id)
new_data = [
    Row(product_id=14, name='Laptop', category='Electronics', price=45000.00, supplier_id=101)
]

df = spark.createDataFrame(new_data)

# Write with mergeSchema option - this will add supplier_id column to the table
df.write \
    .format("delta") \
    .mode("append") \
    .option("mergeSchema", "true") \
    .saveAsTable("dev.bronze.products")

print("✓ Data inserted with schema merge!")
```

#short explain on ACID for  the multiple pipeline


ACID is just like the safety system for the data when multiple actions are performing on the data.

for exp. two pipeline a and b are performing the different task in dataset DS. 
so the acid maintaing the following --
1. Atomicity - nothing or all means either whole data will be insert or nothing.
2. Consistency - remain consistent the rules before and after the data update
3. Isolation - multiple pipeline can  work at same time , no one can see half updates ot each other .
4. Durablity - once the action happen than the data stays even any other issue occures.

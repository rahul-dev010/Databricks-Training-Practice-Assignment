2. Read a CSV with header=True and inferSchema, then read the same file with an explicit StructType
schema; compare the two resulting schemas.


```sql
%python

# df = spark.read.csv("/Volumes/cyntexa_dev/sales/raw/sales2.csv" , header=True, inferSchema=True)
df4 = spark.read.options(header=True, inferSchema=True).csv("/Volumes/cyntexa_dev/sales/raw/sales2.csv")
df4.display()
# df.display()
```

```sql
%python
from pyspark.sql.types import StructType, StructField, IntegerType, DoubleType, StringType

explicit_schema = StructType([
    StructField("order_id",        IntegerType(), True),
    StructField("customer_id",     IntegerType(), True),
    StructField("transaction_id",  IntegerType(), True),
    StructField("product_id",      IntegerType(), True),
    StructField("quantity",        IntegerType(), True),
    StructField("discount_amount", DoubleType(), True),
    StructField("total_amount",    DoubleType(), True),
    StructField("order_date" ,     StringType(), True) 
])

# Read the SAME CSV but with the explicit schema
df = spark.read.csv("/Volumes/cyntexa_dev/sales/raw/sales2.csv" , header=True, schema=explicit_schema)

# df.display()
```

3. Apply one filter transformation followed by one action (count() or show()), and explain in your own
words why nothing ran until the action.

```sql
%python
from pyspark.sql.functions import *
df.filter(col("total_amount")>1000)
    
```

```sql
%python
df.distinct().count()
```

I applied a filter transformation first, but nothing was executed because filter() is a transformation and Spark evaluates transformations lazily. Spark only creates the execution plan and waits. When I used an action like count(), Spark actually executed the filter and processed the data.

Spark works on the concept of lazy evaluation. So, until we use an action like count() or show(), Spark does not actually process or execute the data. It only builds the execution plan for the transformations. When an action is called, Spark executes the required operations and produces the result.

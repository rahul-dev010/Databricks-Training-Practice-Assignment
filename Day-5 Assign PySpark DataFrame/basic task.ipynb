1. Load the messy e-commerce dataset and identify columns containing nulls and duplicate rows using
.filter(), .distinct(), and .dropDuplicates().

```sql
%python
df = spark.read.table("de_dev.bronze.customers")
df = df.distinct().dropDuplicates(subset=["customer_id"]).dropna(how="any", subset=["customer_id", "email"])
df = df.drop("ingestion_timestamp","file_name","file_path","customer_segment")
# df.display()
```

2. Rename at least 3 columns to a consistent naming convention (e.g., snake_case) using
withColumnRenamed.

-> I renamed three columns using withColumnRenamed() to follow a consistent snake_case naming convention

```sql
%python
# df.printSchema()
df = (df
      .withColumnRenamed("name", "c_name")
      .withColumnRenamed("email", "c_email")
      .withColumnRenamed("phone", "c_phone")
    )

df.printSchema()

```

## Genie Agent Curation Checklist for Cyntexa

Before a Genie Agent is trusted to answer executive-facing business questions, the underlying Unity Catalog data should have clear, complete, and reliable metadata.

### 1. Table Descriptions

Each table should have a clear description explaining:

- What the table contains
- Which business process it represents
- The purpose of the table
- How the table is generally used for analysis

For example, a sales table description should explain that it contains customer orders, products, quantities, discounts, and revenue-related information.

### 2. Column Comments

Important columns should have meaningful Unity Catalog comments explaining their business meaning.

The comments should clearly describe:

- What the column represents
- What the values mean
- The unit or currency when applicable
- Whether the column is an ID, date, amount, percentage, or other business field

For example, a `total_amount` column should clearly state whether it represents gross revenue, net revenue, or the final transaction amount.

### 3. Business Metric Definitions

Important business metrics should have a clearly defined calculation method.

Examples include:

- Revenue
- Profit
- Total Sales
- Customer Count
- Order Count
- Growth Rate

The definition should make it clear which columns and calculations should be used so that Genie does not interpret the metric differently.

### 4. Table Relationships

Relationships between important tables should be documented clearly.

For example:

`Customer → Orders → Line Items → Product`

This helps the Genie Agent understand which tables should be joined and which keys should be used.

### 5. Date and Time Definitions

Date columns should have clear comments explaining what they represent.

For example:

- `order_date` → date when the order was placed
- `ship_date` → date when the order was shipped
- `signup_date` → date when the customer registered

This is important for questions involving monthly, quarterly, or yearly trends.

### 6. Data Types and Units

Columns should use appropriate data types and clearly document units.

Examples:

- Revenue → currency
- Quantity → number of units
- Discount → percentage or decimal
- Date fields → DATE or TIMESTAMP

### 7. Sensitive Data and Governance

PII and other sensitive columns should be identified and protected using Unity Catalog governance controls.

The Genie Agent should only expose data that the user is authorized to access.

### 8. Data Quality

Before using the Agent for executive reporting, important tables should be checked for:

- Missing values
- Duplicate records
- Invalid dates
- Incorrect data types
- Unexpected values

### 9. Genie Agent Validation

After the metadata is prepared, the Genie Agent should be tested with common executive questions.

The generated SQL and final answers should be reviewed to confirm that:

- The correct tables are used
- The correct joins are used
- The correct business metrics are calculated
- The results match the expected business logic

### Final Checklist

Before approving an executive-facing Genie Agent:

☑ Table descriptions are available  
☑ Important columns have meaningful comments  
☑ Business metrics have clear definitions  
☑ Table relationships are documented  
☑ Date fields are clearly defined  
☑ Units and data types are correct  
☑ Sensitive data is properly governed  
☑ Data quality has been checked  
☑ Genie responses and generated SQL have been validated  

This checklist can be used at Cyntexa as a standard review process before a Genie Agent is used for executive-facing questions.

## Serverless SQL Warehouse Performance

For our AI/BI dashboard and ad hoc queries, a Serverless SQL Warehouse is a good fit because it is designed for interactive SQL workloads and reduces the need to manage the underlying infrastructure.

### Role of Photon

Photon is the query execution engine that helps SQL queries run efficiently. It improves the performance of analytical queries such as aggregations, joins, and filtering used by our dashboards.

### Role of Predictive I/O

Predictive I/O helps improve data-reading efficiency. It reduces unnecessary data access and helps queries read the required data more efficiently, which can improve response time for dashboard queries.

### Role of Intelligent Workload Management

Intelligent Workload Management helps manage multiple queries and workloads running at the same time. This is useful for dashboards where several users may run different queries or apply filters simultaneously.

### Why Serverless for Our Use Case

Our use case mainly involves AI/BI dashboards, interactive filters, Genie questions, and ad hoc SQL analysis. These workloads require quick query responses and do not require us to manage the underlying compute infrastructure.

Photon provides efficient query execution, Predictive I/O improves data access, and Intelligent Workload Management helps handle concurrent workloads.

Because of these capabilities and the reduced infrastructure management, a Serverless SQL Warehouse is preferred over a classic or Pro warehouse for this interactive dashboard use case.

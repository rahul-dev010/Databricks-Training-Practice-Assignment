```sql
CREATE OR REPLACE FUNCTION cyntexa_dev.sales.maskCutomerId(customer_id STRING)
RETURNS STRING
RETURN 
    CASE WHEN customer_id IS NULL THEN NULL
    ElSE CONCAT('*****',right(customer_id,4))
    END;
```

```sql
SELECT * EXCEPT (customer_id),
cyntexa_dev.sales.maskCutomerId(customer_id) AS masked_customer_id
FROM cyntexa_dev.sales.orders_raw;
```

```sql
--  view joining orders_view with a customers table/view 
-- using the sample data since do not have the customers table 

CREATE OR REPLACE VIEW cyntexa_dev.sales.joining_orders_view AS
SELECT 
c.c_custkey as customer_id,
c.c_name as customer_name,
SUM(O.o_totalprice) AS total_spend
from samples.tpch.customer AS c
JOIN samples.tpch.orders AS o
ON c.c_custkey = o.o_custkey
GROUP BY c.c_custkey, c.c_name;
```

```sql
SELECT *
FROM cyntexa_dev.sales.joining_orders_view
ORDER BY total_spend DESC;
```

-- 1. Use CTAS with read_files() to ingest a CSV file into a managed Delta table.

```sql
CREATE OR REPLACE TABLE cyntexa_dev.sales.sales_csvIngest 
USING DELTA AS
SELECT * FROM
read_files('/Volumes/cyntexa_dev/sales/raw/sales2.csv', format=>'csv',header=>true,inferSchema=>true);

```

-- 2. Ingest a nested JSON file, extracting at least 2 nested fields into top-level columns.
-- ingested the nated json file - customers_nested.json at
path - /Volumes/cyntexa_dev/sales/raw/customers_nested_jsonlines.json 

```sql
CREATE OR REPLACE TABLE cyntexa_dev.default.customer_nestedJson
USING DELTA AS
SELECT 
customer_id,
personal.first_name as first_name,
personal.last_name as last_name,
personal.gender as gender,
contact.email as email,
contact.phone as phone,
contact.address.city as city,
contact.address.state as state,
contact.address.country as country,
employment.company AS company,
employment.department AS department,
employment.designation AS designation
FROM
read_files('/Volumes/cyntexa_dev/sales/raw/customers_nested_jsonlines.json', format=>'json');
```

3. Run DESCRIBE, DESCRIBE EXTENDED, and DESCRIBE DETAIL on your new table and note what unique information each one gives you. 

```sql
DESCRIBE cyntexa_dev.default.customer_nestedjson;
```

```sql
DESCRIBE EXTENDED cyntexa_dev.default.customer_nestedjson;
```

```sql
DESCRIBE DETAIL cyntexa_dev.default.customer_nestedjson;
```
